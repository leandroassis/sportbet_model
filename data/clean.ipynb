# Dataset 1 - Banco dos Dados

```python
dataset1.drop_duplicates(inplace=True)
dataset1.dropna(inplace=True, subset=['ano_campeonato', 'data', 'rodada', 'time_mandante', 'time_visitante', 'gols_mandante', 'gols_visitante']) # Remove rows with missing values in 'ano_campeonato', 'data', 'rodada', 'time_mandante' or 'time_visitante' columns

columns_to_preserve = ['ano_campeonato', 'data', 'rodada', 'time_mandante', 'time_visitante', 'gols_mandante', 'gols_visitante',
                       'estadio', 'arbitro', 'publico', 'publico_max','tecnico_mandante', 'tecnico_visitante', 'colocacao_mandante', 'colocacao_visitante',
                       'valor_equipe_titular_mandante', 'valor_equipe_titular_visitante', 'idade_media_titular_mandante', 'idade_media_titular_visitante']
dataset1.drop(columns=[col for col in dataset1.columns if col not in columns_to_preserve], inplace=True)

print(dataset1.info())
```

```python
import numpy as np
import seaborn as sns

# Define the columns
col1 = 'valor_equipe_titular_mandante'
col2 = 'valor_equipe_titular_visitante'

# Remove NaN values
data1 = dataset1[col1].dropna()
data2 = dataset1[col2].dropna()

# Define quantiles to calculate
step = 0.025
quantiles_to_plot = list(np.arange(0, 0.25 + step, step)) + [0.5, 0.75, 1]

# Calculate quantiles for both datasets
q1 = data1.quantile(quantiles_to_plot)
q2 = data2.quantile(quantiles_to_plot)

# Create a DataFrame for the quantile values for easier reading
quantile_df = pd.DataFrame({
    'Quantile': quantiles_to_plot,
    'Mandante': q1.values,
    'Visitante': q2.values
})

print("Valores dos Quantis:")
print(quantile_df.to_string(index=False))

# Prepare data for violin plot
plot_data = pd.concat([
    pd.DataFrame({'Equipe': 'Mandante', 'Valor': data1}),
    pd.DataFrame({'Equipe': 'Visitante', 'Valor': data2})
])

# Create the plot
plt.figure(figsize=(10, 6))
ax = sns.violinplot(x='Equipe', y='Valor', data=plot_data, inner='box', linewidth=1.5, palette='Set2', hue='Equipe')

# Customizing the plot
ax.set_ylabel('Valor da Equipe')
ax.set_xlabel('')
ax.set_title('Distribuição do Valor das Equipes')
ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
for col in [col1, col2]:
    data = dataset1[col].dropna().sort_values()

    def calculate_distance_to_next(data):
        # Use vectorized operation for efficiency

        distances = (data.iloc[1:].values - data.iloc[:-1].values)/(data.iloc[:-1].values+data.iloc[1:].values)
        max_distance = np.max(distances)
        index_max_distance = np.argmax(distances)
        return max_distance, distances, index_max_distance

    max_distance, distances, index_max_distance = calculate_distance_to_next(data)

    # Main plot
    fig, ax = plt.subplots(figsize=(14, 7))
    
    x_values = range(len(distances))
    ax.bar(x_values, distances, color='skyblue', width=1.0)
    
    ax.set_title(f'Distância entre valores consecutivos para {col}', fontsize=16)
    ax.set_xlabel('Índice dos Valores Ordenados', fontsize=12)
    ax.set_ylabel('Distância', fontsize=12)
    ax.grid(axis='y', linestyle='--', alpha=0.7)
    ax.set_yscale('log')  # Use logarithmic scale for better visualization of distances
    ax.set_ylabel('Distância (Escala Log)', fontsize=12)


    # Inset plot for zoomed view
    ax_inset = fig.add_axes([0.55, 0.55, 0.35, 0.3]) # [left, bottom, width, height]
    
    # Filter data for the inset plot
    zoom_start = 250
    zoom_end = 1000
    ax_inset.bar(np.array(x_values)[zoom_start:zoom_end], distances[zoom_start:zoom_end], color='steelblue', width=1.0)
    
    ax_inset.set_xlabel('Índice')
    ax_inset.set_ylabel('Distância')
    ax_inset.set_yscale('log')
    ax_inset.grid(True, linestyle='--', alpha=0.5)
    
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()
```

```python

from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

all_times = pd.concat([dataset1['time_mandante'], dataset1['time_visitante']]).unique()
all_times_encoded = label_encoder.fit_transform(all_times)

times_dict = dict(zip(all_times, all_times_encoded))

dataset1['time_mandante'] = dataset1['time_mandante'].map(times_dict)
dataset1['time_visitante'] = dataset1['time_visitante'].map(times_dict)

print(dataset1[['time_mandante', 'time_visitante']].head())
print(all_times, all_times_encoded)
```

# Dataset 2 - Football Data

```python
dataset2.drop_duplicates(inplace=True)
dataset2.dropna(inplace=True, subset=['Season', 'HG', 'AG', 'Home', 'Away']) # Remove rows with missing values in 'Season', 'HG', 'AG' or 'Home', 'Away' columns

columns_to_preserve = ['Season', 'Date', 'HG', 'AG', 'Home', 'Away', 'AvgCH', 'AvgCD', 'AvgCA']

dataset2.drop(columns=[col for col in dataset2.columns if col not in columns_to_preserve], inplace=True)

dataset2.head()
```

```python
import unidecode

def normalize_team_name(name):
    """Normalizes team names for consistent matching."""
    if not isinstance(name, str):
        return name
    # Convert to uppercase, remove accents and special characters
    normalized_name = unidecode.unidecode(name).upper()
    # Remove spaces, hyphens, and dots
    normalized_name = normalized_name.replace(' ', '').replace('-', '').replace('.', '')
    return normalized_name

# Create a new dictionary with normalized keys from times_dict
normalized_times_dict = {normalize_team_name(k): v for k, v in times_dict.items()}

# obtidos manualmente
mismatched_teams = {'FIGUEIRENSE': times_dict['Figueirense FC'],
                    'BOTAFOGORJ': times_dict['Botafogo'],
                    'BAHIA': times_dict['EC Bahia'],
                    'VASCO': times_dict['Vasco da Gama'],
                    'FLAMENGORJ': times_dict['Flamengo'],
                    'CORITIBA': times_dict['Coritiba FC'],
                    'VITORIA': times_dict['EC Vitória'],
                    'CRICIUMA': times_dict['Criciúma EC'],
                    'CHAPECOENSESC': times_dict['Chapecoense'],
                    'AVAI': times_dict['Avaí FC'],
                    'JOINVILLE': times_dict['Joinville-SC'],
                    'CEARA': times_dict['Ceará SC'],
                    'BRAGANTINO': times_dict['RB Bragantino'],
                    'CUIABA': times_dict['Cuiabá-MT'],
                    'MIRASSOL': 47,
                    'REMO': 48}

# adiciona as equipes novas ao dicionario de times original
times_dict['Mirassol'] = 47
times_dict['Remo'] = 48

normalized_times_dict_with_mismatched_teams = {**normalized_times_dict, **mismatched_teams}

# Normalize the 'Home' and 'Away' columns in dataset2
dataset2['Home'] = dataset2['Home'].apply(normalize_team_name).map(normalized_times_dict_with_mismatched_teams)
dataset2['Away'] = dataset2['Away'].apply(normalize_team_name).map(normalized_times_dict_with_mismatched_teams)


dataset2[['Home', 'Away']].head()
```

```python
dataset2['Date'] = pd.to_datetime(dataset2['Date'], errors='coerce', dayfirst=True)

dataset2.info()
```

# Combinando os datasets

```python
# Rename columns in dataset2 to match dataset1 for merging
dataset2_renamed = dataset2.rename(columns={
    'Season': 'ano_campeonato',
    'Date': 'data',
    'Home': 'time_mandante',
    'Away': 'time_visitante',
    'HG': 'gols_mandante',
    'AG': 'gols_visitante'
})

# Merge the datasets
# We use an outer merge to keep all rows from both dataset1 and dataset2
# The merge is done on date, home team, and away team
merged_dataset = pd.merge(
    dataset1,
    dataset2_renamed,
    on=['ano_campeonato', 'data', 'time_mandante', 'time_visitante', 'gols_mandante', 'gols_visitante'],
    how='outer'
)

# Display the info of the new merged dataset to check the result
print(merged_dataset.info())
```
